```python
import pandas as pd
import numpy as np
```

```python
def formatarColuna(df : pd.DataFrame, coluna : str):
    df[coluna] = df[coluna].str.replace(r"\s+", " ", regex=True).str.strip().str.title()
```

```python
df = pd.read_csv("dataset/vendedores.csv")
```

```python
formatarColuna(df, "nome")
df["salario"] = df["salario"].fillna(2800.0)
display(df)
```
